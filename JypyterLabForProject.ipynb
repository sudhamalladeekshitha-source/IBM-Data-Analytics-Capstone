```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np

# Clean data
numeric_cols = ['Happiness Score', 'Happiness Rank',
                'Economy (GDP per Capita)', 'Family',
                'Health (Life Expectancy)', 'Freedom',
                'Trust (Government Corruption)', 'Generosity']

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        df[col] = df[col].fillna(df[col].mean())

# Top 10 countries
top10 = df.nsmallest(10, 'Happiness Rank')

# Fig1 - Bar Chart
fig1 = px.bar(top10,
              x='Country',
              y=['Economy (GDP per Capita)', 'Health (Life Expectancy)'],
              title='GDP per Capita and Life Expectancy - Top 10 Countries',
              barmode='group')
fig1.show()

# Fig2 - Correlation Heatmap
cols = ['Economy (GDP per Capita)', 'Family',
        'Health (Life Expectancy)', 'Freedom',
        'Trust (Government Corruption)', 'Generosity',
        'Happiness Score']
corr = df[cols].corr()
fig2 = px.imshow(corr,
                 text_auto=True,
                 title='Correlation Heatmap',
                 width=800, height=600)
fig2.show()

# Fig3 - Scatter Plot
fig3 = px.scatter(df,
                  x='Economy (GDP per Capita)',
                  y='Happiness Score',
                  color='Region',
                  hover_name='Country',
                  title='GDP per Capita vs Happiness Score by Region')
fig3.show()

# Fig4 - Pie Chart
region_happiness = df.groupby('Region')['Happiness Score'].mean().reset_index()
fig4 = px.pie(region_happiness,
              values='Happiness Score',
              names='Region',
              title='Happiness Score by Region')
fig4.show()

# Fig5 - World Map
fig5 = px.choropleth(df,
                     locations='Country',
                     locationmode='country names',
                     color='Economy (GDP per Capita)',
                     hover_name='Country',
                     hover_data=['Health (Life Expectancy)'],
                     title='GDP per Capita by Country',
                     color_continuous_scale='Viridis')
fig5.show()

# Save Dashboard
# Save Dashboard
with open('happiness_dashboard.html', 'w') as f:
    f.write('<html><body>')
    f.write('<h1>World Happiness Report 2016 Dashboard</h1>')
    f.write(fig1.to_html(full_html=False, include_plotlyjs='cdn'))
    f.write(fig2.to_html(full_html=False, include_plotlyjs=False))
    f.write(fig3.to_html(full_html=False, include_plotlyjs=False))
    f.write(fig4.to_html(full_html=False, include_plotlyjs=False))
    f.write(fig5.to_html(full_html=False, include_plotlyjs=False))
    f.write('</body></html>')

print("Dashboard saved successfully!")
```
